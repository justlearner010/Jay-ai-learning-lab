### System of information

#### Types of systems

**1. Non-singular(complete system)**
- Sentences are all informative and non-redundant
- Each sentence contributes new information
- Total information = number of sentence

e.g :
- Dog is black
- Cat is red
- Bird is brown

**2. Singular systems(less informative)**
**(a)Redundant system**
- Repeated information across sentences

e.g:
- Dog is black + Dog is black

**(b)Contradictory systems**
- sentences are conflict
- No consistent solution
